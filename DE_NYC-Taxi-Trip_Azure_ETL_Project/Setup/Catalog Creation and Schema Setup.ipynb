###***Catalog Creation***

```python
%sql
create catalog if not exists nyc_taxitrip
managed location 'abfss://uc-data@stgactnewyorktaxitrip.dfs.core.windows.net/'
comment 'This is a catalog for New York Taxi Trip data';
```

```python
%sql
use catalog nyc_taxitrip;
```

###***Schema Creation***

```python
%sql
create schema if not exists nyc_taxitrip.Bronze;
create schema if not exists nyc_taxitrip.Silver;
create schema if not exists nyc_taxitrip.Gold;
```
