###***Catalog***

```python
catalog=f"nyc_taxitrip"
```

###***Taxi_zone_lookup***

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

brz_path = f"{catalog}.bronze.brz_taxizone_lookup"

df_slv = spark.read.table(brz_path)
df_slv = df_slv.withColumn("Zone",when(col("Zone").isin("NA", "NV"), "N/A").otherwise(col("Zone")))

df_slv.write.option("mergeSchema", "true").mode("overwrite").saveAsTable(f"{catalog}.silver.slv_taxizone_lookup")
df_slv.display()
```

###***yellow_tripdata_2019-01***

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

brz_path = f"{catalog}.bronze.brz_yellow_tripdata_2019_01"

df_slv = spark.read.table(brz_path)
df_slv=df_slv.withColumn('congestion_surcharge', col("congestion_surcharge").cast(DoubleType())).fillna({'congestion_surcharge':0.0})


df_slv.write.option("mergeSchema", "true").mode("overwrite").saveAsTable(f"{catalog}.silver.slv_yellow_tripdata_2019_01")
df_slv.display()
```
