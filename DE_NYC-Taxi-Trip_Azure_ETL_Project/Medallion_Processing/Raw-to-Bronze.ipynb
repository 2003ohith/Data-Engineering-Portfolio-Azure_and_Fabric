###***Taxi_zone_lookup***

```python
raw_path=f"/Volumes/nyc_taxitrip/raw/raw_landing/taxi_zone_lookup/*.parquet"

from pyspark.sql.functions import *
from pyspark.sql.types import *

df_taxizonelookup=spark.read.format("parquet").load(raw_path)
df_taxizonelookup=df_taxizonelookup.withColumn('Ingested_at',current_timestamp())\
    .withColumn("_source_file",col("_metadata.file_path"))

df_taxizonelookup.write.format("delta").mode("overwrite").option("mergeSchema","true").saveAsTable(f"{catalog}.bronze.brz_taxizone_lookup")

df_taxizonelookup.limit(20).display()
```
